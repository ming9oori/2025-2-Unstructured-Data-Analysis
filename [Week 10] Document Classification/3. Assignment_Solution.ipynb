## 1. Import Libraries

*This section is complete - no TODO items here*

````python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, f1_score
import matplotlib.pyplot as plt
import seaborn as sns
from time import time
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline
````

## 2. Load Dataset

*This section is complete - no TODO items here*

````python
from datasets import load_dataset

print("Loading AG News dataset...")
dataset = load_dataset('ag_news')

train_df = pd.DataFrame(dataset['train'])
test_df = pd.DataFrame(dataset['test'])

label_names = {0: 'World', 1: 'Sports', 2: 'Business', 3: 'Sci/Tech'}
train_df['category'] = train_df['label'].map(label_names)
test_df['category'] = test_df['label'].map(label_names)

print(f"Train size: {len(train_df)}")
print(f"Test size: {len(test_df)}")
````

````python
# Use subset for faster training
TRAIN_SIZE = 10000
TEST_SIZE = 2000

train_sample = train_df.sample(n=TRAIN_SIZE, random_state=42)
test_sample = test_df.sample(n=TEST_SIZE, random_state=42)

X_train = train_sample['text']
y_train = train_sample['category']
X_test = test_sample['text']
y_test = test_sample['category']

print(f"Training samples: {len(X_train)}")
print(f"Test samples: {len(X_test)}")
````

---
## 3. Text Vectorization

### TODO 1: Complete the TF-IDF Vectorizer initialization

**Instructions:**  
Fill in the parameters with the following values:
- `max_features`: 5000 (maximum number of features to extract)
- `ngram_range`: (1, 2) (use both unigrams and bigrams)
- `min_df`: 2 (ignore terms that appear in fewer than 2 documents)
- `max_df`: 0.8 (ignore terms that appear in more than 80% of documents)
- `stop_words`: 'english' (remove common English stop words)

**Hint:** Replace each `________` with the appropriate value.

````python
# TODO 1: Complete the TfidfVectorizer initialization
vectorizer_tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.8,
    stop_words='english'
)

print("Vectorizing text data...")
start = time()
X_train_tfidf = vectorizer_tfidf.fit_transform(X_train)
X_test_tfidf = vectorizer_tfidf.transform(X_test)
vectorize_time = time() - start

print(f"Vectorization completed in {vectorize_time:.2f} seconds")
print(f"Vectorized shape: {X_train_tfidf.shape}")
````

### TODO 2: Create Count Vectorizer with the same parameters

**Instructions:**  
Use the same parameters as TODO 1 for the CountVectorizer.

````python
# TODO 2: Complete the CountVectorizer initialization
vectorizer_count = CountVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.8,
    stop_words='english'
)

X_train_count = vectorizer_count.fit_transform(X_train)
X_test_count = vectorizer_count.transform(X_test)

print(f"Count vectorization completed")
print(f"Shape: {X_train_count.shape}")
````

## 4. Naive Bayes Model

### TODO 3: Train Naive Bayes with TF-IDF

**Instructions:**  
Complete the following steps:
1. Create a `MultinomialNB` classifier with `alpha=0.1`
2. Train the model using the `fit()` method with training data
3. Make predictions using the `predict()` method on test data
4. Calculate accuracy using `accuracy_score()`
5. Calculate F1-score using `f1_score()`

**Hint:** Remember that `fit()` needs both X and y, `predict()` only needs X.

````python
# TODO 3: Complete Naive Bayes training with TF-IDF
print("Training Naive Bayes with TF-IDF...")
start = time()

# Create and train the model
nb_tfidf = MultinomialNB(alpha=0.1)
nb_tfidf.fit(X_train_tfidf, y_train)
train_time_nb_tfidf = time() - start

# Make predictions
start = time()
y_pred_nb_tfidf = nb_tfidf.predict(X_test_tfidf)
predict_time_nb_tfidf = time() - start

# Calculate metrics
acc_nb_tfidf = accuracy_score(y_test, y_pred_nb_tfidf)
f1_nb_tfidf = f1_score(y_test, y_pred_nb_tfidf, average='weighted')

print(f"Training time: {train_time_nb_tfidf:.4f} seconds")
print(f"Prediction time: {predict_time_nb_tfidf:.4f} seconds")
print(f"Accuracy: {acc_nb_tfidf:.4f}")
print(f"F1-Score: {f1_nb_tfidf:.4f}")
````

### TODO 4: Train Naive Bayes with Count Vectorizer

````python
# TODO 4: Complete Naive Bayes training with Count Vectorizer
print("Training Naive Bayes with Count Vectorizer...")
start = time()

# Create and train the model
nb_count = MultinomialNB(alpha=0.1)
nb_count.fit(X_train_count, y_train)
train_time_nb_count = time() - start

# Make predictions
start = time()
y_pred_nb_count = nb_count.predict(X_test_count)
predict_time_nb_count = time() - start

# Calculate metrics
acc_nb_count = accuracy_score(y_test, y_pred_nb_count)
f1_nb_count = f1_score(y_test, y_pred_nb_count, average='weighted')

print(f"Training time: {train_time_nb_count:.4f} seconds")
print(f"Prediction time: {predict_time_nb_count:.4f} seconds")
print(f"Accuracy: {acc_nb_count:.4f}")
print(f"F1-Score: {f1_nb_count:.4f}")
````

### Question 2: Vectorization Method  
Compare the performance of TF-IDF vs Count Vectorizer for Naive Bayes. Which one worked better and why? What does this tell you about the importance of feature weighting?

**Answer:**

**Performance comparison:**
- TF-IDF typically achieves 1-3% higher accuracy than Count Vectorizer
- Both perform well, but TF-IDF has a slight edge

**Why TF-IDF works better:**

1. **Feature weighting matters:**
   - Raw counts treat all word occurrences equally
   - TF-IDF down-weights common words that appear in many documents
   - Emphasizes words that are distinctive to specific categories

2. **Example scenario:**
   - Word "said" might appear frequently in all news categories (high count)
   - TF-IDF would assign it lower weight (high document frequency)
   - Word "touchdown" appears mainly in Sports (moderate count)
   - TF-IDF would assign it higher weight (low document frequency, high discrimination power)

3. **Document length normalization:**
   - TF-IDF includes normalization (L2 norm)
   - Prevents longer documents from having artificially higher scores
   - Makes comparison fairer across documents of different lengths

**Importance of feature weighting:**

This comparison demonstrates that **not all features are equally important**:
- Raw frequency can be misleading
- Context matters (how unique is a word across documents?)
- Proper feature engineering (like TF-IDF) can significantly improve model performance
- Even simple transformations can provide meaningful performance gains

For text classification, **TF-IDF is generally preferred** as it better captures the semantic importance of terms.

### Question 4: Training vs Prediction Time
Compare the training and prediction times between Naive Bayes and KNN. Which algorithm is faster for each phase and why? What causes these differences?

**Answer:**

**Observed time patterns:**

**Training Phase:**
- Naive Bayes: ~0.01-0.05 seconds (extremely fast)
- KNN: ~0.001-0.01 seconds (instantaneous)
- **Winner: KNN (technically, but both are very fast)**

**Prediction Phase:**
- Naive Bayes: ~0.01-0.02 seconds (very fast)
- KNN: ~20-50 seconds (significantly slower)
- **Winner: Naive Bayes (by orders of magnitude)**

**Why these differences?**

**Naive Bayes:**
- **Training:** Calculates prior probabilities and likelihood probabilities for each feature
  - Complexity: $O(n \times d)$ where n = samples, d = features
  - Simply counts and calculates probabilities
  - Creates a compact statistical model
  
- **Prediction:** Uses Bayes' theorem with stored probabilities
  - Complexity: $O(c \times d)$ where c = classes, d = features
  - Just multiplies probabilities for each class
  - Very fast computation

**KNN:**
- **Training:** No real training - just stores the training data
  - Complexity: $O(1)$ (constant time)
  - "Lazy learning" - defers computation to prediction time
  - This is why it's so fast
  
- **Prediction:** Must compare test sample with ALL training samples
  - Complexity: $O(n \times d \times k)$ where n = training samples, d = features, k = neighbors
  - For 10,000 training samples × 5,000 features = 50 million comparisons per test sample
  - Extremely computationally expensive
  - Distance calculation in high-dimensional space is costly

**Practical implications:**

1. **For real-time applications:** Naive Bayes is strongly preferred
   - Fast prediction is critical for production systems
   - Can handle millions of predictions per second

2. **For batch processing:** Both could work
   - If processing is done offline, KNN's slow prediction may be acceptable
   - However, Naive Bayes is still more efficient

3. **Scalability concerns:**
   - Naive Bayes scales linearly with data size for both training and prediction
   - KNN prediction time grows linearly with training data size (major bottleneck)
   - With 100,000 training samples, KNN would be 10x slower

**Optimization possibilities:**
- KNN can be optimized using approximate nearest neighbor algorithms (e.g., KD-trees, Ball trees)
- However, these optimizations are less effective in very high dimensions
- Naive Bayes is already optimally efficient for its approach

**Why it helps:**

- **POS tagging:** Nouns/verbs carry more semantic content than articles/prepositions
- **Lemmatization:** Reduces feature space without losing information
- **Custom stopwords:** Removes news-specific noise that doesn't discriminate categories

**Expected improvement:** +2-3% accuracy

**Implementation example:**
```python
import spacy
nlp = spacy.load('en_core_web_sm')

def enhanced_preprocess(text):
    doc = nlp(text)
    
    # Extract entities with weights
    entities = [f"ENT_{ent.label_}_{ent.text}" for ent in doc.ents]
    
    # Keep only meaningful POS tags and lemmatize
    tokens = [token.lemma_ for token in doc 
              if token.pos_ in ['NOUN', 'VERB', 'ADJ']
              and not token.is_stop]
    
    return ' '.join(tokens + entities)
```
